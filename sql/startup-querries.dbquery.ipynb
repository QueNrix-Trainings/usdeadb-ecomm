```sql
CREATE SCHEMA IF NOT EXISTS azdeusdbricks.bronze
MANAGED LOCATION 'abfss://bronze@azdeecomstrgac.dfs.core.windows.net/';

CREATE SCHEMA IF NOT EXISTS azdeusdbricks.silver
MANAGED LOCATION 'abfss://silver@azdeecomstrgac.dfs.core.windows.net/';

CREATE SCHEMA IF NOT EXISTS azdeusdbricks.gold
MANAGED LOCATION 'abfss://gold@azdeecomstrgac.dfs.core.windows.net/';


-------------------------------------------------------------------------

CREATE VOLUME IF NOT EXISTS azdeusdbricks.bronze.raw_files;
CREATE VOLUME IF NOT EXISTS azdeusdbricks.silver.curated_files;
CREATE VOLUME IF NOT EXISTS azdeusdbricks.gold.exports;


--------------------------------------------------------------------------


CREATE SCHEMA IF NOT EXISTS azdeusdbricks.landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS azdeusdbricks.landing.source_files
LOCATION 'abfss://landing@azdeecomstrgac.dfs.core.windows.net/';



SHOW VOLUMES IN azdeusdbricks.landing;
DESCRIBE VOLUME azdeusdbricks.landing.source_files;


GRANT USE CATALOG ON CATALOG azdeusdbricks TO `trainings@quenrix.com`;
GRANT USE SCHEMA  ON SCHEMA  azdeusdbricks.landing TO `trainings@quenrix.com`;


-- read files only (typical for notebooks/jobs that ingest from landing)
GRANT READ VOLUME ON VOLUME azdeusdbricks.landing.source_files TO `trainings@quenrix.com`;

-- write files too (e.g. if Databricks also needs to drop files into landing)
GRANT WRITE VOLUME ON VOLUME azdeusdbricks.landing.source_files TO `trainings@quenrix.com`;




```
